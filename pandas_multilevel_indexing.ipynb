# 🧠 Pandas MultiIndex & Multi-Level Columns


This notebook demonstrates how to use **MultiIndex** (multi-level indexing) in both rows and columns in `pandas`.

Multi-level indexing allows hierarchical organization of data, making it easier to work with panel-like data structures, grouped data, and advanced slicing/grouping operations.


## 🔹 Creating a MultiIndex DataFrame (Rows & Columns)

```python
import pandas as pd
import numpy as np

# Multi-level row index
arrays = [
    ['2023', '2023', '2024', '2024'],
    ['Q1', 'Q2', 'Q1', 'Q2']
]
index = pd.MultiIndex.from_arrays(arrays, names=['Year', 'Quarter'])

# Multi-level column index
cols = pd.MultiIndex.from_tuples([
    ('Revenue', 'Product A'),
    ('Revenue', 'Product B'),
    ('Cost', 'Product A'),
    ('Cost', 'Product B'),
])

data = np.random.randint(100, 500, size=(4, 4))
df = pd.DataFrame(data, index=index, columns=cols)
df
```

## 🔹 Slicing MultiIndex Rows

```python
# Select all data for 2023
df.loc['2023']
```

```python
# Select specific year/quarter
df.loc[('2024', 'Q2')]
```

```python
# Slice across multiple levels
df.loc[('2023',), :]  # Same as above
```

## 🔹 Working with MultiIndex Columns

```python
# Select all Revenue columns
df['Revenue']
```

```python
# Select Revenue for Product A
df['Revenue', 'Product A']
```

```python
# Cross-section using .xs
df.xs('Product A', level=1, axis=1)   # All metrics for Product A
```

```python
df.xs(('2024', 'Q2'), level=('Year', 'Quarter'))
```

## 🔹 Sorting, Swapping & Resetting MultiIndex

```python
# Swap index levels
df_swapped = df.swaplevel()
df_swapped.sort_index()
```

```python
# Sort index explicitly
df.sort_index(level='Quarter', ascending=False)
```

```python
# Reset one or all levels
df.reset_index(level='Quarter')
```

```python
df.reset_index()
```

## 🔹 GroupBy with MultiIndex

```python
# Group by year and aggregate
df.groupby(level='Year').mean()
```

```python
# You can also group by a column level (on axis=1)
df.groupby(level=0, axis=1).sum()  # Sum by 'Revenue' and 'Cost'
```

## ✅ Summary


- MultiIndex enables hierarchical indexing for both rows and columns.
- Use `.loc[]` and `.xs()` for complex selections.
- Combine with `groupby()`, `swaplevel()`, and `sort_index()` for powerful data manipulation.
